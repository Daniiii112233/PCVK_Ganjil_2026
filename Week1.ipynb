```python
import numpy as np  # Library yang digunakan untuk mengelolah matrik/array
import pandas as pd   # Library yang digunakan untuk data berbentuk tabel contoh urls = ["jpg","jpg"]
import cv2 as cv  # Library yang digunakan untuk operasi Visi Komputer dan Pemrograman citra contoh = image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

from google.colab.patches import cv2_imshow   # untuk menampilkan gambar dari google colab
from skimage import io  # fungsinya untuk mengambil modul io dari librarynya skimage kemudian digunakan untuk membaca, menampilkan dan menyimpan gambar
from skimage import transform # fungsinya untuk mengambil modul transform dari librarnya skimage kemudian digunakan untuk melakukan transformasi geometeris pada gambar contohnya garis putih pada soal no 2-4
from PIL import Image # library ini digunakan untuk membaca dan memanipulasi gambarnya contoh nya  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
import matplotlib.pylab as plt  # sedangkan library ini digunakan untuk menampilkan visualisasi data

```

```python
# membuat list untuk menyimpan url dari beberapa image
urls = ["https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQYpQQ-TpIiz-rIlUxcpKba6JiUecKQ4AshbjPnplAUiA&s=10",
        "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRYM3x0K9ptl9fpL1bedh1Lv02qEvfUEXlz9gTzl5GUChsYmiSjrdzRNuKG&s=10"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list

for url in urls:            # Perulangannya
        image = io.imread(url)            # read image
        image = cv.resize(image, (0,0), fx=0.5, fy=0.5)       # resize image to half size
        image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)    # convert color to rgb
        final_frame = cv.hconcat((image,image_2))     # concatenate image
        cv2_imshow(final_frame)             # show image
        print('\n') # spasi pada output gambar
```

```python
tinggi = image_2.shape[0] # Mengambil tinggi gambar (jumlab baris/ pixel vertikal)
lebar = image_2.shape[1]  # Mengambil lebar gambar (jumlab kolom/ pixel horizontal)
print("Resolusi image: tinggi x lebar = ",tinggi, "x" ,lebar) # Menampilkan resolusi gambar dengan menggunakan variable tinggi(vertikal) dan lebar(horizontal)
cv2_imshow(image_2) # Menampilkan gambar
```
