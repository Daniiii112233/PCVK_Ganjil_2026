```python
# Membaca gambar
img = cv2.imread('ark-survival-3840x2160-22103.jpg')

# Konversi ke RGB agar mudah ditampilkan
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Salinan gambar
rb = img_rgb.copy()
gb = img_rgb.copy()

# RB: kanal hijau dinolkan
rb[:, :, 1] = 0

# GB: kanal merah dinolkan
gb[:, :, 0] = 0

# Menampilkan
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(rb)
plt.title("Kombinasi Merah - Biru (RB)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gb)
plt.title("Kombinasi Hijau - Biru (GB)")
plt.axis("off")

plt.show()
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Flip vertikal
flip_vertical = cv2.flip(img_rgb, 0)

# Flip horizontal
flip_horizontal = cv2.flip(img_rgb, 1)

# Flip keduanya
flip_both = cv2.flip(img_rgb, -1)

# Menampilkan
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertical)
plt.title("Flip Vertikal")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Flip Horizontal")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title("Flip Vertikal + Horizontal")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tambah_label_identitas(img, nama="DOKUMENTASI KEGIATAN"):

    hasil = img.copy()

    # Ukuran gambar
    h, w = hasil.shape[:2]

    # Tinggi label proporsional
    bar_height = int(h * 0.12)

    # Ketebalan dan ukuran font proporsional
    thickness = max(1, int(min(w, h) * 0.005))
    font_scale = max(0.5, min(w, h) / 800)

    # Membuat rectangle sebagai label
    cv2.rectangle(
        hasil,
        (0, h - bar_height),
        (w, h),
        (0, 0, 0),
        -1
    )

    # Posisi teks
    x_text = int(w * 0.05)
    y_text = h - int(bar_height * 0.30)

    # Menambahkan teks
    cv2.putText(
        hasil,
        nama,
        (x_text, y_text),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (255, 255, 255),
        thickness,
        cv2.LINE_AA
    )

    return hasil

# Gambar pertama
img1 = cv2.imread('ark-survival-3840x2160-22103.jpg')

hasil1 = tambah_label_identitas(
    img1,
    "DOKUMENTASI KEGIATAN"
)

plt.figure(figsize=(10, 7))
plt.imshow(cv2.cvtColor(hasil1, cv2.COLOR_BGR2RGB))
plt.title("Hasil Studi Kasus 1")
plt.axis("off")
plt.show()
```

```python
def buat_thumbnail(img, nama_toko="TOKO SAYA"):

    # Ukuran gambar asli
    h, w = img.shape[:2]

    # Menentukan ukuran sisi thumbnail
    size = max(h, w)

    # Kanvas putih
    canvas = np.full(
        (size, size, 3),
        255,
        dtype=np.uint8
    )

    # Menghitung posisi tengah
    x_offset = (size - w) // 2
    y_offset = (size - h) // 2

    # Menempatkan gambar di tengah kanvas
    canvas[
        y_offset:y_offset + h,
        x_offset:x_offset + w
    ] = img

    # Ukuran font adaptif
    font_scale = max(0.5, size / 1000)
    thickness = max(1, int(size * 0.005))

    # Posisi teks
    x_text = int(size * 0.05)
    y_text = int(size * 0.95)

    # Background teks
    bar_height = int(size * 0.12)

    cv2.rectangle(
        canvas,
        (0, size - bar_height),
        (size, size),
        (50, 50, 50),
        -1
    )

    # Nama toko
    cv2.putText(
        canvas,
        nama_toko,
        (x_text, y_text),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (255, 255, 255),
        thickness,
        cv2.LINE_AA
    )

    return canvas

img_produk = cv2.imread('ark-survival-3840x2160-22103.jpg')

thumbnail = buat_thumbnail(
    img_produk,
    "TOKO ABDUL"
)

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(thumbnail, cv2.COLOR_BGR2RGB))
plt.title("Thumbnail Marketplace 1:1")
plt.axis("off")
plt.show()

print("Ukuran asli:", img_produk.shape)
print("Ukuran thumbnail:", thumbnail.shape)
```
